```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
URL = 'https://raw.githubusercontent.com/ageron/handson-ml2/master/datasets/housing/housing.csv'
df =pd.read_csv(URL)
df.head()
```

```python
train_set.head()
```

```python
test_set
```

```python
df['median_income'].hist(bins=50, figsize=(9, 6))
plt.show()
```

```python
df['income_cat'] = pd.cut(df['median_income'], bins = [0., 1.5, 3.0, 4.5, 6.0, np.inf], labels =[1, 2, 3, 4, 5])
df['income_cat'].hist(figsize =(9, 6))
plt.show()
```

```python
df.sample(5)
```

```python
stratified_split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in stratified_split.split(df, df['income_cat']):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
strat_train_set['income_cat'].hist(figsize = (9, 6))
plt.show()
```

```python
df['income_cat'].hist(figsize = (9, 6))
```

```python
strat_train_set.drop('income_cat', axis=1, inplace =True)
strat_test_set.drop('income_cat', axis=1, inplace =True)
```

```python
housing = strat_train_set.copy()
housing.head()
```

```python
housing.plot(kind ='scatter', x= 'longitude', y= 'latitude', figsize=(9, 6))
plt.show()
```

```python
housing.plot(
    kind ='scatter',
    x = 'longitude',
    y='latitude',
    alpha =0.4,
    s= housing['population']/100, # nuqta radiusi aholi soniga qarab o'zgaradi
    label = 'population',
    c = 'median_house_value', # nuqta rangi uyning narxiga qarab o'zgaradi
    cmap = 'jet', # ranglar. ko'k -daromadi kam, qizil-daromadi ko'p
    colorbar = True, 
    figsize =(10, 8)

)

plt.show()
    
```

```python
# Select only numeric columns for correlation
numeric_housing = housing.select_dtypes(include=[np.number])

# Calculate correlation with 'median_house_value'
correlations = numeric_housing.corrwith(housing['median_house_value']).sort_values(ascending=False)

# Display the sorted correlations
print(correlations)
```

```python
cols = ['median_house_value', 'median_income', 'total_rooms', 'housing_median_age']
sns.pairplot(housing[cols], height = 5)
plt.show()
```
